# 00 — Đếm dữ liệu RAW

Đọc bốn bộ dữ liệu đã có trong `DATA_RAW` và đếm ảnh cùng nhãn YOLO. Các thư mục train/valid/test của nguồn được gộp để thống kê; chưa tạo lại cách chia dữ liệu.

```python
from pathlib import Path

cwd = Path.cwd().resolve()
project_root = cwd if (cwd / "DATA_RAW").is_dir() else cwd.parents[1]
root = project_root / "DATA_RAW"
datasets = {
    "banknotes_v2": root / "banknotes_v2" / "Banknotes-Recognition-main" / "Dataset_2",
    "money_v1": root / "money_v1",
    "blindtrix_v1": root / "blindtrix_v1",
    "currency_v6": root / "currency_v6",
}
image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
```

```python
rows = []
for dataset_id, dataset_root in datasets.items():
    so_anh = 0
    so_nhan = 0
    for split in ["train", "valid", "test"]:
        image_folder = dataset_root / split / "images"
        label_folder = dataset_root / split / "labels"
        so_anh += sum(1 for file in image_folder.iterdir() if file.is_file() and file.suffix.lower() in image_extensions)
        so_nhan += sum(1 for file in label_folder.iterdir() if file.is_file() and file.suffix.lower() == ".txt")
    rows.append((dataset_id, so_anh, so_nhan))

print(f"{'dataset_id':<16} {'so_anh':>8} {'so_nhan':>8}")
for dataset_id, so_anh, so_nhan in rows:
    print(f"{dataset_id:<16} {so_anh:>8} {so_nhan:>8}")
print(f"{'Tong':<16} {sum(row[1] for row in rows):>8} {sum(row[2] for row in rows):>8}")
```
