# Milestone 3.1 - Inspect Train Image

## Mục tiêu
- Đọc một ảnh PNG từ tập train.
- Kiểm tra cấu trúc dữ liệu ảnh.
- Quan sát shape, dtype, min và max.
- Không thay đổi dữ liệu gốc.

```python
from pathlib import Path
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_dir = Path(r'C:\CS231\DATA\train')

class_dir = Path(train_dir / "500000")
image_path = Path(class_dir / "500000_000090.png")

print("Image path: ", image_path)
```

```python
img = cv2.imread(image_path, cv2.IMREAD_COLOR)
print("Shape:", img.shape)
print("dtype:", img.dtype)
print("Min:", img.min())
print("Max:", img.max())
```

```python
plt.imshow(img)
plt.show()
```

```python
label_from_path = image_path.parent.name
filename = image_path.name
print("Label:", label_from_path)
print("Filename:", filename)
```

```python
from IPython.core.display_functions import display
from collections import Counter
import pandas as pd
size_counts = Counter()
for image_path in train_dir.rglob('*.png'):
    img = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
    height, width = img.shape[:2]
    size_counts[(height, width)] += 1

df_sizes = pd.DataFrame([
    {'height': h, 'width':w, 'count':count} for (h, w), count in size_counts.items()
]).sort_values("count", ascending=False).reset_index(drop=True)

print("Kich thuoc khac nhau:", len(size_counts))
display(df_sizes)
```

```python
from openpyxl.styles.colors import RGB
sample_path = train_dir / "500000" / "500000_000090.png"

# 1. Đọc ảnh bằng cv2.imread
sample_bgr = cv2.imread(str(sample_path), cv2.IMREAD_COLOR)


# 3. Chuyển BGR sang RGB
sample_rgb = cv2.cvtColor(sample_bgr,cv2.COLOR_BGR2RGB)

# 4. Hiển thị ảnh RGB
plt.imshow(sample_rgb)
plt.axis('off')
plt.show()

# 5. In shape của hai mảng
print(sample_rgb.shape)
```
